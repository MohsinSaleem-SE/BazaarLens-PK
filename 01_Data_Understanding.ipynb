Monthly PBS Data Understanding

```python
monthly_df = pd.read_excel(
    r'../Datasets/raw/PBS/SPI-Monthly-Prices-Annex-9.xlsx'
)
```

```python
monthly_df.shape
```

```python
monthly_columns = [
    'Serial_No',
    'Item',
    'Unit',
    'Islamabad',
    'Rawalpindi',
    'Gujranwala',
    'Sialkot',
    'Lahore',
    'Faisalabad',
    'Sargodha',
    'Multan',
    'Bahawalpur',
    'Karachi',
    'Hyderabad',
    'Sukkur',
    'Larkana',
    'Peshawar',
    'Bannu',
    'Quetta',
    'Khuzdar',
    'Sep_2026_Avg',
    'Aug_2026_Avg',
    'Sep_2025_Avg',
    'Pct_Change_Monthly',
    'Pct_Change_Yearly'
]
```

```python
monthly_df.columns = monthly_columns
```

```python
monthly_df.columns
```

```python
monthly_data = monthly_df.iloc[1:].copy()
```

```python
monthly_data.shape
```

```python
monthly_data.head()
```

```python
monthly_data.info()
```

```python
monthly_data['Serial_No'] = monthly_data['Serial_No'].astype(int)
```

```python
numeric_columns = monthly_columns[3:]
```

```python
for col in numeric_columns:
    monthly_data[col] = pd.to_numeric(monthly_data[col], errors='coerce')
```

```python
monthly_data.isna().sum()
```

```python
monthly_data[monthly_data.isna().any(axis=1)]
```

Monthly PBS Dataset — Data Understanding Findings
1. Dataset contains 51 essential food items.
2. Dataset contains prices from 17 cities.
3. Three additional columns provide monthly and yearly average prices and percentage changes.
4. No duplicate records were found.
5. Three missing values were identified for Rice IRRI-6/9 in Gujranwala, Sialkot and Lahore.
6. Numerical columns were converted to appropriate numeric data types.
7. Missing values were retained without replacing them with zero.
