# BazaarLens PK
## Final Insights & Recommendations

This notebook presents the final findings, key insights, and recommendations derived from the analysis of Kaggle and Pakistan Bureau of Statistics (PBS) grocery price datasets.

The objective is to summarize grocery price patterns, city-wise price differences, weekly and monthly price movements, and grocery basket costs across selected cities in Pakistan.

# Section 1 – Project Overview

## 1. Project Overview

BazaarLens PK is a grocery price analysis project focused on understanding grocery prices and price variations across different cities in Pakistan.

The project uses data from two sources:

**1. Kaggle Dataset**
- Contains 350 grocery price records.
- Covers 5 cities in Pakistan.
- Represents grocery prices collected on 28 April 2025.

**2. Pakistan Bureau of Statistics (PBS) Datasets**
- PBS Monthly Dataset: Contains prices of 51 essential items across 17 cities for September 2026.
- PBS Weekly Dataset: Contains five weekly reports from October 2026, covering 51 essential items across 17 cities.

### Project Objectives

1. Analyze grocery prices across different cities.
2. Identify the most expensive grocery items.
3. Examine monthly, yearly, and weekly price changes.
4. Compare average grocery prices across cities.
5. Calculate and compare the cost of a fixed basket of 11 selected grocery items.
6. Identify important price patterns and provide data-driven recommendations.

### Tools and Libraries

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Plotly
- Jupyter Notebook

# Section 2 – Kaggle Dataset: Key Insights

## 2. Kaggle Dataset – Key Insights

The Kaggle dataset contains 350 grocery price records collected on 28 April 2025 from five cities in Pakistan, covering 15 grocery items.

### 2.1 Item-wise Price Insights

1. Eggs (Dozen) recorded the highest average price at Rs. 396.02.
2. Potato ranked second with an average price of Rs. 377.58.
3. Tomato ranked third among the items based on average recorded prices.
4. Moong Dal had the lowest average recorded price at Rs. 243.84.
5. The analysis showed variation in average prices across different grocery items.

### 2.2 City-wise Price Insights

1. Islamabad recorded the highest average grocery price at Rs. 350.33.
2. Lahore ranked second with an average of Rs. 345.65.
3. Peshawar recorded an average price of Rs. 329.24.
4. Quetta recorded an average of Rs. 315.27.
5. Karachi had the lowest average grocery price at Rs. 312.41.

The difference between the highest and lowest city averages was approximately Rs. 37.92.

### 2.3 Data Quality Insights

- The dataset contained 350 records and 15 unique grocery items.
- No missing values or duplicate records were found after cleaning.
- No zero or negative grocery prices were identified.
- The IQR analysis did not identify any price outliers.

### Key Takeaway

The Kaggle dataset highlights differences in grocery prices across five Pakistani cities. Islamabad had the highest average recorded grocery price, while Karachi had the lowest.

**Limitation:** The dataset represents a single collection date (28 April 2025), so it cannot be used to analyze historical price trends or inflation over time. Also, different grocery items have different measurement units.

# Section 4: PBS Weekly Dataset – Key Insights

## 4. PBS Weekly Dataset – Key Insights

The PBS Weekly dataset contains five weekly reports from October 2026, covering 51 essential items across 17 cities in Pakistan.

### 4.1 Weekly Price Trend Insights

The analysis of selected grocery items revealed the following price movements:

1. Chicken Farm Broiler (Live) recorded the highest increase of approximately 18.02% during the five weeks.
2. Wheat Flour Bag (20 Kg) increased by approximately 9.42%.
3. Milk fresh (Un-boiled) showed a relatively small increase of approximately 0.50%.
4. Rice Basmati Broken (Average Quality) showed a slight decrease of approximately 0.80%.

Chicken and wheat flour showed notable price increases during October 2026.

### 4.2 City-wise Price Insights

1. Islamabad recorded the highest average item price at Rs. 804.21.
2. Rawalpindi ranked second with an average of Rs. 777.89.
3. Karachi recorded an average of Rs. 760.18.
4. Peshawar recorded an average of Rs. 752.86.
5. Sukkur had the lowest average item price at Rs. 698.99.

The difference between the highest and lowest city averages was approximately Rs. 105.22.

### 4.3 Data Quality Insights

- The dataset combined five weekly reports dated 1, 8, 15, 22, and 29 October 2026.
- Each weekly report covered 51 essential items.
- Missing values for some items and city-level prices were retained during cleaning.
- The dataset included minimum, average, and maximum prices for different cities.

### Key Takeaway

Chicken and wheat flour experienced the most noticeable price increases among the four selected items. Islamabad had the highest average item price, while Sukkur had the lowest among the 17 cities.

**Limitation:** The analysis covers only five weeks and four selected items for weekly trend comparisons. It should not be treated as a long-term inflation analysis.

# Section 6 – Grocery Basket Final Insights

## 6. Grocery Basket – Final Insights

The grocery basket analysis compared the total cost of a fixed basket containing 11 selected essential grocery items across 17 cities in Pakistan.

### 6.1 Most Expensive Cities

1. Islamabad had the highest basket cost at Rs. 8,226.68.
2. Karachi ranked second at Rs. 8,098.75.
3. Peshawar ranked third at Rs. 7,971.08.
4. Rawalpindi ranked fourth at Rs. 7,928.89.
5. Quetta ranked fifth at Rs. 7,924.62.

### 6.2 Cheapest Cities

1. Faisalabad had the lowest basket cost at Rs. 7,164.49.
2. Sargodha ranked second lowest at Rs. 7,183.53.
3. Lahore ranked third lowest at Rs. 7,202.21.
4. Multan ranked fourth lowest at Rs. 7,295.39.
5. Gujranwala ranked fifth lowest at Rs. 7,524.54.

### 6.3 Overall Basket Cost Difference

- Highest basket cost: Islamabad – Rs. 8,226.68.
- Lowest basket cost: Faisalabad – Rs. 7,164.49.
- Difference: Approximately Rs. 1,062.19.

### Key Takeaway

Islamabad had the most expensive selected grocery basket, while Faisalabad had the cheapest. The difference of approximately Rs. 1,062.19 highlights the variation in the cost of the selected grocery basket across cities.

**Limitation:** This analysis is based on a fixed basket of 11 selected items using PBS September 2026 prices. It does not represent the total cost of all groceries or every household's actual spending.

# Section 8 – Final Conclusion & Project Limitations

## 8. Final Conclusion & Project Limitations

### 8.1 Final Conclusion

BazaarLens PK analyzed grocery prices across different cities in Pakistan using Kaggle and Pakistan Bureau of Statistics (PBS) datasets.

The project identified several important findings:

1. Islamabad recorded the highest average prices in all three datasets.
2. Chicken and wheat flour showed notable weekly price increases during October 2026.
3. Grocery basket analysis showed that Islamabad had the highest selected basket cost, while Faisalabad had the lowest.

5. City-wise analysis highlighted variations in grocery prices across Pakistan.

Overall, this project demonstrates how Python and data analysis techniques can help identify grocery price patterns, compare cities, and understand changes in essential commodity prices.

### 8.2 Project Limitations

1. **Kaggle Dataset:** Contains prices collected on a single date, 28 April 2025, so it cannot show historical price trends.
2. **PBS Monthly Dataset:** Represents monthly prices for September 2026 and does not provide a long-term monthly history.
3. **PBS Weekly Dataset:** Covers only five weeks of October 2026, which is insufficient for long-term trend analysis.
4. **Selected Items:** Weekly trend analysis focuses on only four selected grocery items.
5. **Grocery Basket:** The basket contains only 11 selected items and does not represent every household's grocery expenses.
6. **Different Measurement Units:** Grocery items have different units, so their prices should not always be compared directly.
7. **Missing Values:** Some missing PBS price values were retained during cleaning rather than estimated.
